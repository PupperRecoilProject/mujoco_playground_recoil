```python
# ===================================================================
# 儲存格 1: 沿用【最終升級版】地形生成器類別
# ===================================================================
import numpy as np
import matplotlib.pyplot as plt
from scipy.ndimage import gaussian_filter
from mpl_toolkits.mplot3d import Axes3D

class TiledTerrainGenerator:
    """
    【最終升級版】地形生成器
    - 新增真實比例 3D 視覺化功能
    """
    def __init__(self, terrain_params, tile_physical_size=5.0, terrain_weights=None):
        self.params = terrain_params
        self.tile_physical_size = tile_physical_size
        default_weights = {"flat": 0.5, "random_noise": 0.2, "steps": 0.15, "slope": 0.15}
        if terrain_weights: default_weights.update(terrain_weights)
        total_weight = sum(default_weights.values())
        self.terrain_weights = {k: v / total_weight for k, v in default_weights.items()}
        self.terrain_types = list(self.terrain_weights.keys())
        self.terrain_probabilities = list(self.terrain_weights.values())
        self.generators = { "flat": self.generate_flat, "random_noise": self.generate_random_noise, "steps": self.generate_steps, "slope": self.generate_slope }
        print("✅ (儲存格 1) 【最終升級版】地形生成器類別已定義。")
        self.taper_mask = None

    def _create_taper_mask(self, rows, cols):
        self.taper_mask = np.outer(np.hanning(rows), np.hanning(cols))

    def generate_flat(self, tile_rows, tile_cols): return np.zeros((tile_rows, tile_cols))
    def generate_random_noise(self, tile_rows, tile_cols): return (np.random.rand(tile_rows, tile_cols) - 0.5) * self.params.get("random_noise_height", 0.05)
    def generate_steps(self, tile_rows, tile_cols):
        single_step_height = self.params.get("single_step_height", 0.04)
        hfield = np.zeros((tile_rows, tile_cols)); num_steps = np.random.randint(3, 8)
        step_width = tile_rows // num_steps
        for i in range(num_steps): hfield[i*step_width:(i+1)*step_width, :] = (i + 1) * single_step_height
        return hfield
    def generate_slope(self, tile_rows, tile_cols):
        angle_deg = self.params.get("slope_angle_deg", 5.0)
        max_height = (self.tile_physical_size / 2.0) * np.tan(np.deg2rad(angle_deg))
        slope = np.zeros((tile_rows, tile_cols)); direction = np.random.choice(['x', 'y', '-x', '-y'])
        x = np.linspace(-1, 1, tile_cols); y = np.linspace(-1, 1, tile_rows); X, Y = np.meshgrid(x, y)
        if direction == 'x': slope = (X + 1) / 2
        elif direction == 'y': slope = (Y + 1) / 2
        elif direction == '-x': slope = (-X + 1) / 2
        elif direction == '-y': slope = (-Y + 1) / 2
        return slope * max_height

    def generate_tiled_map(self, grid_dim, tile_dim, smoothing_sigma):
        grid_rows, grid_cols = grid_dim; tile_rows, tile_cols = tile_dim
        self._create_taper_mask(tile_rows, tile_cols)
        full_map = np.zeros((grid_rows * tile_rows, grid_cols * tile_cols))
        for r in range(grid_rows):
            for c in range(grid_cols):
                terrain_type = np.random.choice(self.terrain_types, p=self.terrain_probabilities)
                tile = self.generators[terrain_type](tile_rows, tile_cols)
                if terrain_type != "flat": tile -= np.mean(tile)
                tile *= self.taper_mask
                row_start, row_end = r * tile_rows, (r + 1) * tile_rows
                col_start, col_end = c * tile_cols, (c + 1) * tile_cols
                full_map[row_start:row_end, col_start:col_end] = tile
        if smoothing_sigma > 0: full_map = gaussian_filter(full_map, sigma=smoothing_sigma)
        min_val, max_val = np.min(full_map), np.max(full_map)
        if max_val == min_val: return full_map, full_map
        return full_map, (full_map - min_val) / (max_val - min_val)

    def visualize_terrain_2d(self, terrain_data, title="Grayscale Heightmap (2D View)"):
        plt.figure(figsize=(8, 8)); plt.imshow(terrain_data, cmap='gray', origin='lower')
        plt.colorbar(label='Normalized Height'); plt.title(title); plt.show()

    def visualize_terrain_3d_interactive(self, terrain_data, physical_size, title="Interactive 3D View", vertical_exaggeration=1.0):
        """
        【最終升級版】使用 Plotly 生成可互動的 3D 預覽圖。
        - 新增 vertical_exaggeration 參數以控制 Z 軸拉伸。
        """
        import plotly.graph_objects as go

        rows, cols = terrain_data.shape
        phys_y, phys_x = physical_size
        x_m = np.linspace(0, phys_x, cols)
        y_m = np.linspace(0, phys_y, rows)
        
        print("🚀 正在生成可互動的 Plotly 3D 圖表...")
        if vertical_exaggeration != 1.0:
            print(f"   ⚠️  注意: Z 軸已手動拉伸 {vertical_exaggeration} 倍以凸顯細節。")
        else:
            print("   ✅ 當前為 1:1 真實物理比例。")

        fig = go.Figure(data=[go.Surface(x=x_m, y=y_m, z=terrain_data, colorscale='Viridis', colorbar=dict(title='Height (m)'))])
        
        z_range = np.max(terrain_data) - np.min(terrain_data)
        if z_range < 1e-6: z_range = max(phys_x, phys_y) * 0.1

        fig.update_layout(
            title=title,
            width=900, height=700,
            scene=dict(
                xaxis_title='X (meters)',
                yaxis_title='Y (meters)',
                zaxis_title='Z (meters)',
                aspectmode='manual',
                # 【核心】應用垂直誇大係數
                aspectratio=dict(x=phys_x, y=phys_y, z=z_range * vertical_exaggeration)
            )
        )
        fig.show()
```

```python
# ===================================================================
# 儲存格 2: 設定您的完整參數
# ===================================================================
GRID_DIM = (2, 2)
TILE_PHYSICAL_SIZE = 5.0
TILE_RESOLUTION = 512 # 使用 512 以加快預覽，需要時可改回 1024
SMOOTHING = 8.0 # 建議稍大一點的值以獲得更柔和的拼接

# 【關鍵】在這裡定義所有地形的參數
terrain_features = {
    "slope_angle_deg": 1.0,       # 您指定的斜坡角度
    "single_step_height": 0.02,   # 階梯的單階高度
    "random_noise_height": 0.003   # 崎嶇地形的起伏
}

# 【關鍵】在這裡設定您想要的地形組合和機率
terrain_distribution = {
    "flat": 0.25,
    "slope": 0.1,
    "steps": 0.25,
    "random_noise": 0.5
}

generator = TiledTerrainGenerator(
    terrain_params=terrain_features, 
    tile_physical_size=TILE_PHYSICAL_SIZE,
    terrain_weights=terrain_distribution
)
print("✅ (儲存格 2) 地形生成器實例已建立，所有地形功能均已啟用。")
```

```python
# ===================================================================
# 儲存格 3: 執行生成
# ===================================================================
full_map, normalized_map = generator.generate_tiled_map(grid_dim=GRID_DIM, tile_dim=(TILE_RESOLUTION, TILE_RESOLUTION), smoothing_sigma=SMOOTHING)
total_x_meters = GRID_DIM[1] * TILE_PHYSICAL_SIZE
total_y_meters = GRID_DIM[0] * TILE_PHYSICAL_SIZE
print("\n--- 地圖資訊 ---")
print(f"地圖物理尺寸: {total_y_meters:.1f}m x {total_x_meters:.1f}m")
print(f"地形最高點: {np.max(full_map):.3f} m, 最低點: {np.min(full_map):.3f} m")
```

```python
# ===================================================================
# 儲存格 4: 互動式 3D 視覺化
# ===================================================================
print("\n📏 正在呈現【可互動的、真實比例】3D 預覽圖...")
generator.visualize_terrain_3d_interactive(full_map, physical_size=(total_y_meters, total_x_meters))
```
